```python
M     = 15                    # Parameter to be changed
nD    = 2                     # white-noise case, parameter to be changed
mu    = 1e-3                  # must satisfy 0 < mu < 2/((M+1)*var(x)), will be sweeped as a parameter
h     = [np.zeros(M + 1)]     # weights; h[n][k]
```

**The feedback loop**
- Create loop
  - produce `e[n] = x[n] - i_processed[n]`
  - calculate `h[k,n+1] = h[k,n] + mu * e[n] * x[n - delay - k]`
  - calculate `i_processed[n+1] = h[k,n] convolved with x[n]`
  - plot mean-square error (formula from class slides)

```python
for n in range(nD + M, len(x)):
        u = x[n - nD - np.arange(M + 1)]         # [x[n-nD], ..., x[n-nD-M]]
        e[n] = x[n] - h[-1] @ u                  # e[n] = x[n] - ihat[n]
        h.append(h[-1] + mu * e[n] * u)          # LMS update -> h[:, n+1]
```
